```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
page_cnt = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_cnt}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore...")
        break

    with open(f"scraped_data/quotes{page_cnt}.html" , "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"downloaded data from page {page_cnt}")

    page_cnt = page_cnt+1
```

```python
life_quotes = []

for page_cnt in range(1, 11):

    with open(f"scraped_data/quotes{page_cnt}.html", "r", encoding="utf-8") as f:
        html_content = f.read()

    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")

    for q in all_quotes:
        all_tags = []

        for tag in q.select(".tags .tag"):
            all_tags.append(tag.get_text())

        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()

            life_quotes.append([text, author])

# print(life_quotes)
```

```python
import pandas as pd

df = pd.DataFrame(life_quotes, columns=["Quote", "Author"])

df.to_csv("cleaned_data/life_quotes.csv", index=False)

print("Rows in DataFrame:", len(df)) 

print("CSV file created successfully")
        
```
